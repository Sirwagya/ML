```python
booking = pd.read_csv("../Datasets/bookings.csv")
booking['customer'] = booking['customer'].str.strip().str.upper()
booking['booking_id'] = booking['booking_id'] - 1000
booking['day'] = booking['day'].str[:3].str.upper()
print (booking.head())

xd = booking.groupby('customer').agg(
    is_Regular = (("customer", "count"))
)

xd.reset_index()
xd['is_Regular'] = xd['is_Regular'] > 1
print (xd)
```
