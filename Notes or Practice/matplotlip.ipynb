```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
titanic = pd.read_csv('../Datasets/tested.csv')
mpg = pd.read_csv('../Datasets/mpg.csv')
```

```python
plt.figure(figsize=(6,6))
sns.countplot(data= titanic, x='Survived')
plt.title("AGE VS FARE")

print (titanic['Fare'].median(),titanic['Fare'].describe()['mean'].round(2))
```

```python
plt.figure(figsize=(9,6))
sns.histplot(data= mpg, x='mpg' , kde = True, bins = 25)
plt.title("MPG DISTRIBUTION")
plt.show()
mpg['mpg'].describe()['mean']
```

```python
plt.figure(figsize=(9,6))
sns.countplot(data= mpg, x='origin' )
plt.title("Origin")
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.histplot(data= mpg, x='mpg', y = 'horsepower' , bins = 25)
plt.title("Origin")
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.boxplot(data= mpg, x='mpg', y = 'origin')
plt.title("Origin")
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(data= titanic, x='Age', y = 'Fare' , hue = 'Sex', palette = 'Set2')
plt.title("Origin")
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.boxplot(data= mpg, x='origin')
plt.title("Origin")
plt.show()
plt.figure(figsize=(5,5))
sns.boxplot(data= mpg, x='cylinders')
plt.title("Origin")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data= mpg, x='weight', y = 'mpg' , hue = 'origin')
plt.title("Origin")
plt.show()
```
