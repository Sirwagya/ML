## np.random 
- Seed 
- Rand

```python
np.random.seed(2)
print (np.random.rand(10))
print (np.random.randn(100))
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots()   
np.random.seed(50)
x = np.sort(np.random.randn(500) )
ax.plot(range(500), x**2, marker='.')
plt.show()                        
```

```python
np.random.seed(50)
x = np.random.randn(5000)  # don't sort

fig, ax = plt.subplots()   
ax.hist(x, bins=30, density=True)
plt.show()
```

```python
lst = np.arange (23)
y = np.random.choice(lst,25)
print (*y)
```

### Basic Statistics

```python
arr = np.arange(6).reshape(2,3)
print (arr)
print (arr.min())
print (arr.min(axis=0), "column min")
print (arr.min(axis=1), "row min")
```

```python
arr = np.arange(6,12).reshape(2,3)
print (arr)
print (arr.sum())
print (arr.sum(axis=0), "column sum")
print (arr.sum(axis=1), "row sum")
```

```python
print (*(*np.median(arr, axis=0).astype('int64'), *np.median(arr, axis=1).astype('int64')))
```
